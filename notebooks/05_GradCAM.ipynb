# 05 — Explainable AI: Grad-CAM

Visualizes which regions of a chest X-ray drove the model's prediction.

```python
import sys
sys.path.append('..')
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from config import CONFIG
from src.inference import InferenceEngine
%matplotlib inline
```

```python
model_path = CONFIG.paths.best_model_path
if model_path.exists():
    engine = InferenceEngine(model_path, class_names=CONFIG.data.class_names, image_size=CONFIG.data.image_size)
else:
    engine = None
    print(f'No trained model found at {model_path}. Run train.py first.')
```

## Run inference + Grad-CAM on a sample test image

```python
if engine is not None:
    sample_dir = CONFIG.paths.test_dir / 'PNEUMONIA'
    files = list(sample_dir.iterdir()) if sample_dir.exists() else []
    if files:
        result = engine.predict_path(files[0])
        fig, axes = plt.subplots(1, 3, figsize=(14, 5))
        axes[0].imshow(result['preprocessed_image']); axes[0].set_title('Preprocessed'); axes[0].axis('off')
        axes[1].imshow(result['heatmap'], cmap='jet'); axes[1].set_title('Grad-CAM heatmap'); axes[1].axis('off')
        axes[2].imshow(result['overlay']); axes[2].set_title(f"Overlay — {result['prediction']} ({result['confidence']*100:.1f}%)"); axes[2].axis('off')
        plt.show()
    else:
        print('No test images found under dataset/test/PNEUMONIA/.')
```
