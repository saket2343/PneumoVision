# 02 — Preprocessing

Demonstrates the deterministic preprocessing chain from `src/preprocessing.py`: resize -> RGB conversion -> CLAHE -> normalization (Phase 2).

```python
import sys
sys.path.append('..')
import matplotlib.pyplot as plt
from config import CONFIG
from src.preprocessing import load_image, resize_image, apply_clahe, normalize, preprocess_image
%matplotlib inline
```

## Pick a sample image

```python
sample_dir = CONFIG.paths.train_dir / 'PNEUMONIA'
sample_files = list(sample_dir.iterdir()) if sample_dir.exists() else []
print(f'Found {len(sample_files)} candidate images')
sample_path = sample_files[0] if sample_files else None
sample_path
```

## Step-by-step pipeline visualization

```python
if sample_path:
    raw = load_image(sample_path)
    resized = resize_image(raw, CONFIG.data.image_size)
    clahe_img = apply_clahe(resized)
    final = normalize(clahe_img)

    fig, axes = plt.subplots(1, 4, figsize=(16, 4))
    for ax, img, title in zip(
        axes, [raw, resized, clahe_img, final],
        ['Original', 'Resized', 'CLAHE', 'Normalized [0,1]']
    ):
        ax.imshow(img)
        ax.set_title(title)
        ax.axis('off')
    plt.show()
else:
    print('No sample image available yet — add data under dataset/train/PNEUMONIA/.')
```

## One-call preprocessing (used identically at train/eval/inference time)

```python
if sample_path:
    out = preprocess_image(raw, size=CONFIG.data.image_size, use_clahe=True)
    print(out.shape, out.dtype, out.min(), out.max())
```
